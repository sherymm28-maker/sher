```python

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average(self):
        total = 0

        for grade in self.grades:
            total += grade

        return total / len(self.grades)


s = Student("Sara", [85, 90, 76])
print(s.name)
print(s.average())

t = Student("Omar", [70, 80])
print(t.name)
print(t.average())
```
