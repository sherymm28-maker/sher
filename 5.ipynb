```python
def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average(self):
        total = 0
        count = 0

        for grade in self.grades:
            total += grade
            count += 1

        if count == 0:
            return 0

        return total / count


s = Student("Sara", [85, 90, 76])
print(s.name)

assert s.name == "Sara"
assert abs(s.average() - 83.67) < 0.01

t = Student("Omar", [70, 80])
assert t.name == "Omar"
assert abs(t.average() - 75.0) < 0.01

print("A1: ممتاز")
```
