```python
grades = {"math": 88, "english": 76}

grades["math"] = 92
grades["physics"] = 79

del grades["english"]

best = max(grades.values())

print(grades, best)

assert grades == {"math": 92, "physics": 79}
assert best == 92

print("A3: ممتاز")




```

```python
student = ("ليلى", 21, 92.5)

name, age, score = student

print(name, age, score)

pairs = [("أحمد", 77), ("سارة", 85)]

first_name, first_score = pairs[0]

print(first_name, first_score)

assert name == "ليلى" and age == 21 and score == 92.5
assert first_name == "أحمد" and first_score == 77

print("A4: ممتاز")
```

```python
cart = []

cart.append("Latte")
cart.append("Croissant")
cart.append("Tea")

cart.remove("Croissant")
cart.insert(0, "Water")
cart.pop()
cart.sort()

print(cart, len(cart))

assert cart == ["Latte", "Water"]
assert len(cart) == 2

print("A1: ممتاز")
```
